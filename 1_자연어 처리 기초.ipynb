# Corpus
Corpus(말뭉치)는 자연어 처리에서 분석하거나 학습하기 위해 수집한 전체 텍스트 데이터의 집합


### Doucument
- Corpus를 구성하는 각각의 텍스트 데이터
- Document라고 해서 반드시 긴 문서를 의미하는 것은 아님
- 예) 뉴스 기사 하나, 리뷰 하나, 댓글 하나, 문장 하나 등

# **3. Tokenization**
- 텍스트를 모델이 처리할 수 있는 작은 단위인 Token으로 나누는 과정
- [토큰 계산](https://platform.openai.com/tokenizer)


### 토큰
- 토큰은 모델이 처리하는 텍스트의 작은 단위
- 하나의 토큰은 단어, 부분적인 단어, 공백이나 구두점일 수 있음
- 토큰은 반드시 단어 하나를 의미하지는 않음

### Vocabulary
- Vocabulary는 Tokenizer가 사용하는 토큰들의 집합
- 각 토큰에는 고유한 정수 ID가 할당됨
- 정수 ID 자체는 언어적인 의미가 전혀 없음

### 허깅 페이스
허깅 페이스(Hugging Face)는 자연어 처리뿐 아니라 이미지, 음성, 영상, 멀티모달 모델을 공유하고 사용하는 데 필요한 여러 도구와 서비스를 제공하는 AI 생태계이자 회사 이름 

````python
import torch
````

````python
if torch.cuda.is_available():
    DEVICE = torch.device('cuda')
elif torch.backends.mps.is_available():
    DEVICE = torch.device('mps')
elif torch.xpu.is_available():
    DEVICE = torch.device('xpu')
else:
    DEVICE = torch.device('cpu')

print(DEVICE)
````

````python
text = "나는 자연어를 공부한다"
tokens = text.split()
# tokens
# 초반에 스페셜 토큰이 먼저 나옴
# 0은 padding을 넣는 단어를 많이 넣음
# unk : 알지 못하는 
vocab = {"<PAD>":0, "<UNK>":1, "나는":2, "자연어를":3, "공부한다":4}
token_ids = [vocab.get(token, vocab['<UNK>']) for token in tokens]

input_ids = torch.tensor(token_ids, dtype=torch.long)

print("원문: ", text)
print("Token: ", tokens)
print("Token ID: ", token_ids)
print("PyTorch Tensor: ", input_ids)
print("dtype: ", input_ids.dtype)
````

````python
from transformers import AutoTokenizer
````

````python
# 허깅 페이스 허브의 "klue/bert-base" Repository에서 Tokenizer 설정과 Voca를 불러옴
# tokenizer 설정하고 vocabulary 가져옴
# klue : 한국어가 제일 잘 학습된 말뭉치
# bert : 구글에서 만든 모델 이름(제미나이 초창기 버전)
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")
````

````python
text = '자연어 처리를 공부합니다.'
tokens = tokenizer.tokenize(text) # tokenizer 내부에 형태소 분석기가 존재
print("Token: ",tokens)
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print("token ID: ", token_ids)
````

````python
print("Vocabulary 크기: ", tokenizer.vocab_size)
print("[PAD] ID: ", tokenizer.pad_token_type_id)
print("[UNK] ID: ", tokenizer.unk_token_id)
print("[CLS] ID: ", tokenizer.cls_token_id)
print("[SEP] ID: ", tokenizer.sep_token_id)
````

### OOV
- OOV(Out-Of-Vocabulary)는 Voca에 존재하지 않는 Token이 입력되는 문제(예.신조어,오타,처음 보는 단어)
- 과거의 단어 단위 Voca에서는 처음 보는 단어를 \<UNK\>와 같은 Unknown Token으로 처리하는 경우가 많았음
- 현대 NLP에서는 Subword Tokenization을 이용하여 이 문제를 크게 줄임

````python
vocab = {"<PAD>":0, "<UNK>":1, "나는":2, "파이토치를":3, "공부한다":4}
tokens = {"나는","트랜스포머를","공부한다"}
token_ids = [vocab.get(token, vocab['<UNK>']) for token in tokens]
print(token_ids)
````

### Token ID
Tokenization이 끝나면 각각의 Token을 Vocabulary에 정의된 정수 ID로 변환
```
"오늘 날씨가 좋다" > Tokenization > [Token1, Token2, Token3 ...] > Token ID > [102, 235, 303, ...]
```
